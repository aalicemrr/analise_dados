```python
import pandas as pd
```

```python
caminho_arquivo = "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

dados = pd.read_excel(caminho_arquivo, sheet_name="Municípios", header=2)

print(dados.columns.tolist())
dados.head()
```

```python
colunas_necessarias = [
    "UF",
    "NOME DO MUNICÍPIO",
    "População 2010 (Alterações de Limites até 2022)1",
    "População Censo 2022"
]

dados_limpos = dados[colunas_necessarias].dropna()
dados_limpos.head()
```

```python
populacao_estados = dados_limpos.groupby("UF")[[
    "População 2010 (Alterações de Limites até 2022)1",
    "População Censo 2022"
]].sum().reset_index()

populacao_estados["Diferenca_2010_2022"] = (
    populacao_estados["População Censo 2022"] - populacao_estados["População 2010 (Alterações de Limites até 2022)1"]
)

populacao_estados = populacao_estados.sort_values("Diferenca_2010_2022", ascending=False)
```

```python
populacao_estados
```

```python
populacao_estados.to_csv("crescimento_estados_alice.csv", sep=";", index=False)
```

```python
populacao_municipios = dados_limpos.copy()

populacao_municipios["Diferenca_2010_2022"] = (
    populacao_municipios["População Censo 2022"] - populacao_municipios["População 2010 (Alterações de Limites até 2022)1"]
)
```

```python
populacao_municipios = populacao_municipios.sort_values("Diferenca_2010_2022", ascending=False)
```

```python
populacao_municipios.head(10)
```

```python
populacao_municipios.to_csv("crescimento_municipios_alice.csv", sep=";", index=False)
```
