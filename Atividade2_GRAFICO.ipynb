```python
import pandas as pd
```

```python
path = "Tabela 1.1.1.xlsx"
```

```python
bruto = pd.read_excel(path, engine="openpyxl", header=None)
```

```python
indicador_1 = bruto.iloc[8:41].copy()
indicador_1.columns = [
    "uf_regiao",
    "total",
    "total_branca",
    "total_preta_parda",
    "homem_branca",
    "homem_preta_parda",
    "mulher_branca",
    "mulher_preta_parda",
]
indicador_1 = indicador_1.reset_index(drop=True)
```

```python
for col in indicador_1.columns[1:]:
    indicador_1[col] = pd.to_numeric(indicador_1[col], errors="coerce")
```

```python
indicador_1
```

```python
media_brasil = indicador_1.loc[indicador_1["uf_regiao"] == "Brasil", "total"].item()
media_brasil
```

```python
grandes_regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
dados_estados = indicador_1[
    ~indicador_1["uf_regiao"].isin(["Brasil"] + grandes_regioes)
]
```

```python
acima_media = dados_estados.loc[
    dados_estados["total"] > media_brasil,
    ["uf_regiao", "total"]
].sort_values("total", ascending=False)

print(f"MÉDIA DO BRASIL: {media_brasil:.2f} horas/semana")
print(f"ESTADOS ACIMA DA MÉDIA: {len(acima_media)}")
acima_media
```

```python
abaixo_media = dados_estados.loc[
    dados_estados["total"] < media_brasil,
    ["uf_regiao", "total"]
].sort_values("total", ascending=False)

print(f"MÉDIA DO BRASIL: {media_brasil:.2f} horas/semana")
print(f"ESTADOS ABAIXO DA MÉDIA: {len(abaixo_media)}")
abaixo_media
```

```python
linha_brasil = indicador_1.loc[indicador_1["uf_regiao"] == "Brasil"].iloc[0]

media_homens = (
    linha_brasil["homem_branca"] + linha_brasil["homem_preta_parda"]
) / 2

media_mulheres = (
    linha_brasil["mulher_branca"] + linha_brasil["mulher_preta_parda"]
) / 2

aumento_percentual = (media_mulheres - media_homens) / media_homens * 100

print(f"HOMENS: {media_homens:.2f} h/semana")
print(f"MULHERES: {media_mulheres:.2f} h/semana")
print(f"DIFERENÇA PERCENTUAL: {aumento_percentual:.1f}%")
```

```python
perc_branca = (
    linha_brasil["mulher_branca"] / linha_brasil["homem_branca"] - 1
) * 100

perc_preta_parda = (
    linha_brasil["mulher_preta_parda"] / linha_brasil["homem_preta_parda"] - 1
) * 100

print(f"MULHERES BRANCAS: +{perc_branca:.1f}%")
print(f"MULHERES PRETAS OU PARDAS: +{perc_preta_parda:.1f}%")
```

```python
linha_mg = indicador_1.loc[indicador_1["uf_regiao"] == "Minas Gerais"].iloc[0]
diferenca_mg = linha_mg["mulher_preta_parda"] - linha_mg["mulher_branca"]

linha_brasil = indicador_1.loc[indicador_1["uf_regiao"] == "Brasil"].iloc[0]
diferenca_brasil = linha_brasil["mulher_preta_parda"] - linha_brasil["homem_branca"]

linha_amapa = indicador_1.loc[indicador_1["uf_regiao"] == "Amapá"].iloc[0]
diferenca_amapa = linha_amapa["mulher_preta_parda"] - linha_amapa["mulher_branca"]
```

```python
print("RESUMO DAS RESPOSTAS")
print(f"MINAS GERAIS: {diferenca_mg:.1f} hora(s)")
print(f"BRASIL: {diferenca_brasil:.1f} hora(s)")
print(f"AMAPÁ: {diferenca_amapa:.1f} hora(s)")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
resumo_regioes = indicador_1[
    indicador_1["uf_regiao"].isin(grandes_regioes)
].copy()

resumo_regioes["HOMENS"] = (
    resumo_regioes["homem_branca"] + resumo_regioes["homem_preta_parda"]
) / 2

resumo_regioes["MULHERES"] = (
    resumo_regioes["mulher_branca"] + resumo_regioes["mulher_preta_parda"]
) / 2

dados_grafico = resumo_regioes.melt(
    id_vars=["uf_regiao"],
    value_vars=["HOMENS", "MULHERES"],
    var_name="SEXO",
    value_name="HORAS"
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=dados_grafico,
    x="uf_regiao",
    y="HORAS",
    hue="SEXO"
)
plt.title("AFAZERES DOMÉSTICOS POR REGIÃO E SEXO")
plt.xlabel("GRANDE REGIÃO")
plt.ylabel("HORAS POR SEMANA")
plt.tight_layout()
plt.show()
```
