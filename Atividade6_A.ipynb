```python
import pandas as pd

dados_2012 = pd.read_csv(
    "Tabela5-sem_emprego_2012.csv",
    sep=";",
    decimal=","
).rename(columns={
    "Desocupados - homens (2012 T1)": "homens_2012",
    "Desocupados - mulheres (2012 T1)": "mulheres_2012"
})

dados_2026 = pd.read_csv(
    "Tabela5-sem_emprego_2026.csv",
    sep=";",
    decimal=","
).rename(columns={
    "Desocupados - homens (2026 T1)": "homens_2026",
    "Desocupados - mulheres (2026 T1)": "mulheres_2026"
})
```

```python
chaves = ["Sigla", "Código", "Estado"]

comparacao = pd.merge(
    dados_2012,
    dados_2026,
    on=chaves,
    how="inner"
)

comparacao["mudanca_homens"] = (
    comparacao["homens_2026"] - comparacao["homens_2012"]
)

comparacao["mudanca_mulheres"] = (
    comparacao["mulheres_2026"] - comparacao["mulheres_2012"]
)

comparacao
```

```python
chaves = ["Sigla", "Código", "Estado"]

comparacao = pd.merge(
    dados_2012,
    dados_2026,
    on=chaves,
    how="inner"
)

comparacao["mudanca_homens"] = (
    comparacao["homens_2026"] - comparacao["homens_2012"]
)

comparacao["mudanca_mulheres"] = (
    comparacao["mulheres_2026"] - comparacao["mulheres_2012"]
)

comparacao
```

```python
bruto = pd.read_excel(
    "Tabela 1.1.1.xls",
    sheet_name=0,
    header=None,
    skiprows=8,
    nrows=33
)

horas_estado = bruto.iloc[:, [0, 1, 4, 5, 6, 7]].copy()

horas_estado.columns = [
    "Estado",
    "media_horas_2022",
    "horas_homens_brancos",
    "horas_homens_pretos_pardos",
    "horas_mulheres_brancas",
    "horas_mulheres_pretas_pardas"
]

horas_estado = horas_estado[
    horas_estado["Estado"].isin(comparacao["Estado"])
].copy()

colunas_numericas = horas_estado.columns[1:]

horas_estado[colunas_numericas] = horas_estado[
    colunas_numericas
].apply(pd.to_numeric, errors="coerce")

horas_estado
```

```python
resultado_alice = comparacao.merge(
    horas_estado,
    on="Estado",
    how="inner"
)

resultado_alice = resultado_alice.sort_values(
    "Estado"
).reset_index(drop=True)

resultado_alice.to_csv(
    "geral_alice.csv",
    sep=";",
    decimal=",",
    index=False
)

print(f"{len(resultado_alice)} estados foram unidos")

resultado_alice
```
