```python
import pandas as pd
```

```python
caminho = "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
```

```python
dados_brutos = pd.read_excel(caminho, header=None, engine="openpyxl")
print(dados_brutos.shape)
print(dados_brutos.head(10))
```

```python
dados = dados_brutos.iloc[3:36].reset_index(drop=True)

dados.columns = [
    "localidade",
    "total_geral",
    "masculino_geral",
    "feminino_geral",
    "total_tecnologia",
    "masculino_tecnologia",
    "feminino_tecnologia",
    "total_servicos_saude",
    "masculino_servicos_saude",
    "feminino_servicos_saude",
]

print(dados.head(10))
```

```python
colunas_numericas = [
    "total_geral",
    "masculino_geral",
    "feminino_geral",
    "total_tecnologia",
    "masculino_tecnologia",
    "feminino_tecnologia",
    "total_servicos_saude",
    "masculino_servicos_saude",
    "feminino_servicos_saude",
]

for coluna in colunas_numericas:
    dados[coluna] = pd.to_numeric(dados[coluna], errors="coerce")

print(dados.dtypes)
```

```python
grandes_regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
print(grandes_regioes)
```

```python
dados_regioes = dados[dados["localidade"].isin(grandes_regioes)].reset_index(drop=True)
print(dados_regioes)
```

```python
tabela_superior = dados_regioes[["localidade", "masculino_geral", "feminino_geral"]].copy()
tabela_superior.columns = ["Região", "Homens", "Mulheres"]
print(tabela_superior)
```

```python
print(tabela_superior.describe())
```

```python
tabela_tecnologia = dados_regioes[["localidade", "masculino_tecnologia", "feminino_tecnologia"]].copy()
tabela_tecnologia.columns = ["Região", "Homens", "Mulheres"]
print(tabela_tecnologia)
```

```python
print(tabela_tecnologia.describe())
```

```python
tabela_saude = dados_regioes[["localidade", "masculino_servicos_saude", "feminino_servicos_saude"]].copy()
tabela_saude.columns = ["Região", "Homens", "Mulheres"]
print(tabela_saude)
```

```python
print(tabela_saude.describe())
```

```python
tabela_superior.to_excel("formacao_superior_regioes_alice.xlsx", index=False)
tabela_tecnologia.to_excel("areas_tecnologicas_regioes_alice.xlsx", index=False)
tabela_saude.to_excel("educacao_saude_regioes_alice.xlsx", index=False)
```

```python
print("Arquivos Excel criados com sucesso:")
print("- formacao_superior_regioes_alice.xlsx")
print("- areas_tecnologicas_regioes_alice.xlsx")
print("- educacao_saude_regioes_alice.xlsx")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
linha_brasil = dados[dados["localidade"] == "Brasil"].iloc[0]

resumo_brasil = pd.DataFrame({
    "AREA": [
        "Graduação total",
        "Tecnologia / STEM",
        "Saúde e educação"
    ],
    "HOMENS": [
        linha_brasil["masculino_geral"],
        linha_brasil["masculino_tecnologia"],
        linha_brasil["masculino_servicos_saude"]
    ],
    "MULHERES": [
        linha_brasil["feminino_geral"],
        linha_brasil["feminino_tecnologia"],
        linha_brasil["feminino_servicos_saude"]
    ]
})

formacao_grafico = resumo_brasil.melt(
    id_vars=["AREA"],
    value_vars=["HOMENS", "MULHERES"],
    var_name="SEXO",
    value_name="TOTAL_PESSOAS"
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=formacao_grafico,
    x="AREA",
    y="TOTAL_PESSOAS",
    hue="SEXO"
)
plt.title("BRASIL - FORMAÇÃO ACADÊMICA POR SEXO")
plt.xlabel("ÁREA DE FORMAÇÃO")
plt.ylabel("TOTAL DE PESSOAS")
plt.tight_layout()
plt.show()
```
